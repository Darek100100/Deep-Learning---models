## Accuracy - Dokładność klasyfikacji

$$
\text{Accuracy} = \left( \frac{\text{Correct Predictions}}{\text{Total Predictions}} \right) \times 100\%
$$



```python
import numpy as np

y_ture = np.array([1, 0, 1, 1, 0, 1])
y_pred = np.array([0, 0, 1, 1, 0, 1])

def accuracy(y_true, y_pred):
  correct = 0
  for i in range(len(y_true)):
    if y_true[i] == y_pred[i]:
      correct += 1
  return correct / len(y_true) * 100
```

```python
accuracy(y_ture, y_pred)
```

## Entropia rozkładu prawdopodobieństwa




$$
Entropy = - \sum_{i=1}^{n} p_i \log (p_i)
$$


```python
import numpy as np
from math import log, e

def entropy(labels, base=None):
    n_labels = len(labels)
    if n_labels <= 1:
        return 0

    value, counts = np.unique(labels, return_counts=True)
    probs = counts / n_labels
    n_classes = np.count_nonzero(probs)

    if n_classes <= 1:
        return 0

    ent = 0.

    base = e if base is None else base

    for i in probs:
        ent -= i * log(i, base)

    return ent

labels = [1, 3, 5, 2, 3, 5, 3, 2, 1, 3, 4, 5]

# Obliczenie entropii
result = entropy(labels, base=2)
print("Entropia:", result)
```

## Binarna Entropia Krzyżowa

```python
def binary_crossentropy(y_true, y_pred):
  y_pred = np.clip(y_pred, 0.0000001, 1-0.0000001)
  return -y_true * np.log(y_pred) - (1 - y_true) * np.log(1 - y_pred)
```

```python
y_true = np.array([1, 0, 1, 1, 0, 1, 0])
y_pred = np.array([0, 0, 1, 1, 0, 1, 1])

binary_crossentropy(y_true, y_pred)
```

## Kategoryczna Entropia Krzyżowa

```python
def categorical_crossentropy(y_true, y_pred):
  y_pred = np.clip(y_pred, 0.0000001, 1-0.0000001)
  return -np.sum (y_true * np.log(y_pred))
```

```python
y_ture = np.array([1, 0, 0, 0, 0])
y_pred = np.array([0.4, 0.3, 0.05, 0.05, 0.2])

categorical_crossentropy(y_ture, y_pred)
```

```python
y_true = np.array([1, 0, 0, 0, 0]),
y_pred = np.array([0.98, 0.01, 0.0, 0.0, 0.01])

categorical_crossentropy(y_true, y_pred)
```
