## Funkcje Aktywacji
### ReLU Function
$f(x) = max (x,0)$

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()
```

```python
def max_relu(x):
  return max(x, 0)
```

```python
for i in[-10, -5, 0, 5, 10]:
  print(max_relu(i))
```

```python
data = np.arange(-5, 5, 0.1)
max_relu_data = [max_relu(x) for x in data]

plt.plot(data, max_relu_data)
plt.title('ReLU Function')
```
