## Implementacja sieci neuronowej

```python
import numpy as np

np.random.seed(0)

X = np.array([[1.0, 0.7]])
y_true = np.array([1.80])
```

```python
def initialize_parameters(input_size, hidden_size, output_size):
    W1 = np.random.randn(input_size, hidden_size)
    W2 = np.random.randn(hidden_size, output_size)
    return W1, W2

# Propagacja w przód
def forward_propagation(X, W1, W2):
    H = np.dot(X, W1)
    y_pred = np.dot(H, W2)
    return H, y_pred

def calculate_error(y_true, y_pred):
  return y_true - y_pred

def backpropagation(X, W1, W2, learining_rate = 0.01, iters = 1000, precision = 0.00000001):

  H, y_pred = forward_propagation(X, W1, W2)

  for i in range(iters):
    error = calculate_error(y_true, y_pred)
    W2 = W2 - learining_rate * error * H.T
    W1 = W1 - learining_rate * error * X.T * W2.T

    _, y_pred = forward_propagation(X, W1, W2)

    print('Iter {}, y_pred: {}, error: {}'.format(i, y_pred, calculate_error(y_true, y_pred)))

    if abs(error) < precision:
      break
  return W1, W2

def predict(X, W1, W2):
  _, y_pred = forward_propagation(X, W1, W2)
  return y_pred


def build_model():
  ##inicjalizacja wag
  W1, W2 = initialize_parameters(2, 2, 1)
  ##propagacja wsteczna
  W1, W2 = backpropagation(X, W1, W2)

  model = {
      'W1': W1,
      'W2': W2
  }
  return model

model = build_model()


```
